<h2>2.4&nbsp;&nbsp;&nbsp;&nbsp;Checking For Missing Data</h2>

<p>Missing data must be identified before modeling. If missing values are not handled correctly, the model may fail during training or produce unreliable results.</p>


```python
df.isna().sum().sort_values(ascending=False).to_frame(name='Missing Values')
```

<p><strong>Interpretation:</strong> The missing value check shows which columns require imputation. In the production workflow, imputation is performed inside the scikit-learn pipeline after the train/test split to avoid data leakage.</p>


<hr/>
<h2>2.5&nbsp;&nbsp;&nbsp;&nbsp;Understanding Machine Status Distribution</h2>

<p>The target distribution is important because an imbalanced target can make accuracy misleading. In predictive maintenance, the fault class is usually less frequent but more important to detect.</p>


```python
target_distribution(df)
```

<p><strong>Interpretation:</strong> The target distribution shows that normal machine status is much more common than abnormal status. Therefore, model evaluation should include metrics such as balanced accuracy, recall, F1 score, ROC AUC, and average precision rather than relying only on accuracy.</p>
