<h3>3.4.1&nbsp;&nbsp;&nbsp;&nbsp;Imputing Missing Numerical Data</h3>

<p>I will begin by imputing missing numerical data first. For numerical data, we will be using the mean strategy as this will mitigate the problems that will be caused by outliers as mean will never be an outlier and will also not skew the graph in anyway. Subsequently, I will also display the newly imputed dataframe so as to verify that the code has been successfully executed and there are no longer any 'nan' values within any numeric data columns.</p>

```python
# Selection of all numeric data columns in the DataFrame
numeric_cols = df.select_dtypes(include='number')

# Create an instance of SimpleImputer with mean strategy
mean_imputer = SimpleImputer(strategy='mean')

# Fit and transform the numeric columns using the mean imputer
numeric_cols_imputed = mean_imputer.fit_transform(numeric_cols)

# Replace the original numeric columns with the imputed values in the DataFrame
df[numeric_cols.columns] = numeric_cols_imputed

# Display the first 10 rows of the imputed DataFrame
df.head(10).style.background_gradient(cmap="Blues")
```

<p>From the output of the DataFrame as shown in the cell above, I can confidently verify that the code that performs the imputation of the missing values for Numeric Data has been successfully executed. I can also verify from the dataframe that there are no 'nan' values within the first 10 data rows as compared to previously where there was a missing data in the index 6 under both the data column 'Process T (C)' and 'Rotation Speed (rpm)'. In conclusion, this proves and shows that the imputation of the missing numerical data was conducted and executed successfully.</p>

<hr/>
<h3>3.4.2&nbsp;&nbsp;&nbsp;&nbsp;Imputing Missing Categorical Data</h3>

<p>I will now be proceed to impute the missing Categorical Data. As Categorical Data is not a Numerical Data, there will not be a mean to use. Therefore, I will be using the mode (most_frequent) to replace the missing data. As the mode is the data with the highest frequency, by adding imputing data with reference to mode, there will not be much changes as the mode will still remain the same as the frequency of the mode will only increase and it will still keep its status of being the mode.</p>

```python
# Selection of all categorical data columns in the DataFrame
categorical_cols = df.select_dtypes(include='object')

# Create an instance of SimpleImputer with the most frequent strategy
mode_imputer = SimpleImputer(strategy='most_frequent')

# Fit and transform the categorical columns using the mode imputer
categorical_cols_imputed = mode_imputer.fit_transform(categorical_cols)

# Replace the original categorical columns with the imputed values in the DataFrame
df[categorical_cols.columns] = categorical_cols_imputed

# Display the first 10 rows of the imputed DataFrame
df.head(10).style.background_gradient(cmap="Blues")
```

As there are no missing categorical data for the first 10 data rows, I am unable to provide evidence that the data has been successfully imputed. <h2>4.3&nbsp;&nbsp;&nbsp;&nbsp;Handling Outliers</h2>

<p>With all the facts from the previous visualisations conducted, I will now be proceeding to handle the Outliers. The observations form the visualisations that I have made have been listed below.</p>
<br/>
<p><strong><u>Rotation Speed (rpm)'s Outlier</u></strong></p>
<ul>
    <li>High Frequency of Outlier Closer to Maximum (Boxplot)</li>
    <li>Includes Extreme Values (e.g. 2750+ rotation per minute)</li>
    <li>Extremely Postively-skewed (Histogram)</li>
</ul>

<p><strong><u>Torque (Nm)'s Outlier</u></strong></p>
<ul>
    <li>High Frequency of Outlier that are Extremely Close to Minimum and Maximum (Boxplot)</li>
    <li>Includes Slightly Extreme Values (e.g. 5Nm, 75Nm)</li>
    <li>Slightly Postively-skewed and Slightly Negatively-skewed (Histogram)</li>
</ul>
<br/>
<p>With all these evaluations from the visualisations, I have decided to set up an <strong><u>Upper Fence and Lower Fence</u></strong> with respect to their respective Inter-Quartile Range. This will be the most fair and safe approach as it removes the extreme values but also not remove all the Outliers which might be genuie data, especially Outliers which has high frequency. The formula which I will be utilising to make this ammendment is shown below,</p><br/>

$$IQR = Q3 - Q1$$
$$\text{Lower Fence} = Q1 - (1.5 \times IQR)$$
$$\text{Upper Fence} = Q3 + (1.5 \times IQR)$$
<br/>
<p>With reference to the formula, I will be establishing the Lower and Upper Fence in the cell below.</p>


```python
# Calculating IQR for Rotation Speed
Q1_rotation_speed = df["Rotation Speed (rpm)"].quantile(0.25)
Q3_rotation_speed = df["Rotation Speed (rpm)"].quantile(0.75)
IQR_rotation_speed = Q3_rotation_speed - Q1_rotation_speed

# Calculating IQR for Torque
Q1_torque = df["Torque (Nm)"].quantile(0.25)
Q3_torque = df["Torque (Nm)"].quantile(0.75)
IQR_torque = Q3_torque - Q1_torque

# Filtering Outliers for Rotation Speed
df = df[(df["Rotation Speed (rpm)"] >= Q1_rotation_speed - 1.5 * IQR_rotation_speed) & (df["Rotation Speed (rpm)"] <= Q3_rotation_speed + 1.5 * IQR_rotation_speed)]

# Filtering Outliers for Torque
df = df[(df["Torque (Nm)"] >= Q1_torque - 1.5 * IQR_torque) & (df["Torque (Nm)"] <= Q3_torque + 1.5 * IQR_torque)]

# Displaying the top 10 data rows
df.head(10).style.background_gradient(cmap="Blues")
```

<p>With reference to the DataFrame output produced by the cell above, I can verify that the Lower Fence and Upper Fence has been established, removing any Outliers which are beyond the 2 fence. Subsequently, I will be using another Boxplot to verify that the Outliers have been removed successfully. <h2>4.6&nbsp;&nbsp;&nbsp;&nbsp;Creation of New Features</h2>

In this sub-section, I will be creating 2 new features which I believe will make the model more accurate and be beneficial. During my research, I stumbled upon 2 formulas which can be used to create 2 new features. I will be adding Temperature Gradient and Power Indicator into my dataset. These 2 features can be calculated with the data that I have within the dataset. These 2 features will give me even more insight into what caused a machine to be abnormal. The new features also describe the relationship between the features used for calculation. The formulas for the 2 new features are shown below.
<br/>
$$
\text{Temperature Gradient} = T_{\text{process}} - T_{\text{ambient}}
$$
<br/>
$$
\text{Power Indicator} = \frac{\text{Torque}}{\text{Rotational Speed}}
$$
<br/>
The formulas are also capable of describing the relationship between the features that are used to calculate the new features:

- **Temperature Gradient**:
    - If Process Temperature **increases** while Ambient Temperature **remains the same**, Temperature Gradient will **increase**.
    - If Process Temperature **remains the same** while Ambient Temperature **increases**, Temperature Gradient will 
**decrease**.

- **Power Indicator**:
    - If Torque **increases** while Rotational Speed **remains the same**, Power Indicator will **increase**.
    - If Torque **remains the same** while Rotational Speed **increases**, Power Indicator will **decrease**.

Upon stating the formula and the relationship, I will proceed with creating the new features in the cell below.

```python
# Creating Temperature Gradient Column
df['Temperature Gradient'] = df['Process T (C)'] - df['Ambient T (C)']

# Creating Power Indicator Data Column
df['Power Indicator'] = df['Torque (Nm)'] / df['Rotation Speed (rpm)']

# Viewing of Updated Data Table
df.head(10).style.background_gradient(cmap="Blues")
```

<h2>8.1&nbsp;&nbsp;&nbsp;&nbsp;Learning Curve Plotting Function</h2>

<p>I will be plotting the learning curve multiple times to detect for overfitting and underfitting or ideal fitting. Therefore, I will be utilising a function such that I do not have to retype the code everytime when plottng the learning curve. A learning curve is capable of showing the following results and the action plan that needs to be done to rectify the issue (if any).</p>
<br/>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>Cross-Validation Score matches the Training Score at the final point.</li>
    <li>Hyperparameter have been tuned successfully.</li>
</ul>
<p><strong><u>Overfitting</u></strong></p>
<ul>
    <li>Cross-Validation Score fall short of the Training Score.</li>
    <li>Hyperparameter needs to be adjusted accordingly by reducing the model's complexity.</li>
</ul>
<p><strong><u>Underfitting</u></strong></p>
<ul>
    <li>Training Score fall short of the Cross-Validation Score.</li>
    <li>Hyperparameter needs to be adjusted accordingly by increase the model's complexity.</li>
</ul>
<br/>
<p>With the potential results and action plan recorded, I will proceed to implement the Learning Curve Plotting function in the cell below.</p>

```python
# Improved Learning Curve Function for Subplot
def plot_learning_curve_subplot(estimator, title, X, Y, ax, ylim = None, cv = None, n_jobs = 1, train_sizes = np.linspace(0.1, 1.0, 5)):
    sns.set(style = "whitegrid")
    ax.set_title(title, fontsize = 12)
    ax.set_xlabel("Training examples", fontsize = 10)
    ax.set_ylabel("Score", fontsize = 10)
    if ylim is not None:
        ax.set_ylim(*ylim)

    # Compute learning curve data
    train_sizes, train_scores, test_scores = learning_curve(
        estimator, X, Y, cv = cv, n_jobs = n_jobs, train_sizes = train_sizes
    )
    
    # Calculate mean and standard deviation
    train_scores_mean = np.mean(train_scores, axis = 1)
    train_scores_std = np.std(train_scores, axis = 1)
    test_scores_mean = np.mean(test_scores, axis = 1)
    test_scores_std = np.std(test_scores, axis = 1)

    # Fill between for train and test standard deviations
    ax.fill_between(train_sizes, train_scores_mean - train_scores_std,
                    train_scores_mean + train_scores_std, alpha = 0.1, color = "blue")
    ax.fill_between(train_sizes, test_scores_mean - test_scores_std,
                    test_scores_mean + test_scores_std, alpha = 0.1, color = "orange")

    # Plot the mean scores for training and testing sets
    sns.lineplot(x = train_sizes, y = train_scores_mean, color = "blue", marker = 'o', label = "Training Score", ax = ax)
    sns.lineplot(x = train_sizes, y = test_scores_mean, color = "orange", marker = 'o', label = "Cross-Validation Score", ax = ax)

    # Highlight the final points for emphasis
    ax.scatter(train_sizes[-1], train_scores_mean[-1], color = 'blue', s = 50, edgecolor = 'k', linewidth = 1.5, label = 'Final Training Point')
    ax.scatter(train_sizes[-1], test_scores_mean[-1], color = 'orange', s = 50, edgecolor = 'k', linewidth = 1.5, label = 'Final Cross-Validation Point')
    ax.legend(loc = "best", fontsize = 8)
    ax.grid(True)

# Code Execution Confirmation
print("Function Successfully Defined")
```

```python
def plot_confusion_matrix_subplot(estimator, X_train, Y_train, ax, title):
    # Predict using the estimator
    Y_pred = estimator.predict(X_train)

    # Generate confusion matrix
    cm = confusion_matrix(Y_train, Y_pred)

    # Plot the confusion matrix using Seaborn
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False, linewidths=1, linecolor='black', ax=ax, annot_kws={"size": 50})
    ax.set_title(title, fontsize=22)
    ax.set_xlabel('Predicted Labels', fontsize=20)
    ax.set_ylabel('True Labels', fontsize=20)
    ax.tick_params(axis='both', which='major', labelsize=20)

# Code Execution Confirmation
print("Function Successfully Defined")
```

<p>The last part of the Preliminary Training of Models is to evaluate the respective scoring of the model. These scoring includes accuracies, balanced accuracies and F1 Score. I will also be colour coding the data which are not ideal and data that are ideal in red and green respectively. After the scoring evaluation, I may need to compare certain models with each other to evaluate their individual performance.</p>

```python
# Define scoring metrics for cross-validation
scoring_metrics = {
    'accuracy': make_scorer(accuracy_score),
    'balanced_accuracy': 'balanced_accuracy',
    'recall': make_scorer(recall_score, average = 'macro'),
    'f1': make_scorer(f1_score, average = 'macro'),
    'roc_auc': make_scorer(roc_auc_score, multi_class = 'ovr')
}

# List to hold cross-validation results for all models
results = []

# Perform cross-validation on each model and store the results
for model_name, model in classification_models:
    cv = ShuffleSplit(n_splits = 10, test_size = 0.2, random_state = 28)
    scores = cross_validate(
        model,
        X_train,
        Y_train,
        scoring = scoring_metrics,
        cv = cv,
        return_train_score = True,
        n_jobs =- 1
    )
    
    # Append the average score for each model to the results
    avg_scores = {metric: scores[metric].mean() for metric in scores}
    avg_scores['model'] = model_name
    results.append(avg_scores)

# Convert the list of results to a DataFrame, setting the model names as index
outputs_df = pd.DataFrame(results)
outputs_df.set_index('model', inplace = True)

# Function to highlight thresholds for every single data box
def highlight_thresholds(df):
    # Function to apply color based on value range
    def apply_highlight(val, threshold_good, threshold_acceptable):
        if (val < threshold_acceptable):
            return 'background-color: red; color: white'
        elif (val < threshold_good):
            return 'background-color: yellow; color: black'
        else:
            return 'background-color: green; color: white'

    # Apply styling for various metrics using thresholds
    styled_df = df.style.applymap(
        lambda x: apply_highlight(x, 0.85, 0.70), subset=['test_accuracy', 'train_accuracy']
    ).applymap(
        lambda x: apply_highlight(x, 0.75, 0.60), subset=['test_balanced_accuracy', 'train_balanced_accuracy']
    ).applymap(
        lambda x: apply_highlight(x, 0.70, 0.50), subset=['test_recall', 'train_recall', 'test_f1', 'train_f1']
    ).applymap(
        lambda x: apply_highlight(x, 0.75, 0.60), subset=['test_roc_auc', 'train_roc_auc']
    )

    return styled_df

# Apply the new highlighting to your DataFrame
highlighted_df = highlight_thresholds(outputs_df)

# Display the highlighted DataFrame
highlighted_df
```

<h3>9.1.3&nbsp;&nbsp;&nbsp;&nbsp;Random Forest Classifier Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best hyperparameter for the Random Forest Classifier, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the Random Forest Classifier. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good accuracy score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>

<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid for GridSearchCV
grid_param_rfc = {
    'randForestClassifier__n_estimators': [30, 50, 75],
    'randForestClassifier__max_depth': [3, 5, 7],
    'randForestClassifier__min_samples_split': [30, 40],
    'randForestClassifier__min_samples_leaf': [20, 25],
    'randForestClassifier__bootstrap': [True],
    'randForestClassifier__max_features': ['log2']
}

# Create a GridSearchCV instance using KFold
grid_search_rfc = GridSearchCV(
    estimator=pipeline_rfc,
    param_grid=grid_param_rfc,
    cv=kf,
    verbose=1,
    n_jobs=-1
)

# Fit GridSearchCV to the training data
grid_search_rfc.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_rfc.cv_results_)

# Sort the DataFrame to push the best parameter row to the top
grid_sorted_results_df = grid_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

As only one model can be chosen between the Random Forest Classifier and Decision Tree, I will be comparing the best Hyperparameter of the Decision Tree with Random Forest Classifier to evaluate if the difference in accuracy is worth the transparency. <hr/>
<h3>9.2.1&nbsp;&nbsp;&nbsp;&nbsp;Decision Tree's Initiation and Pipeline</h3>

<p>In order to start training the data for Decision Tree, I will firstly initiate it and also set up the pipeline which will scale the training data. I will also need to ensure that the random state is always 28 which is standardised throughout all the random state value. Scaling of data will also reduce biasness, ensure the feature is dimensionless and improve the model more accurate and effective as extreme values are removed.</p>

```python
# Initiating Decision Tree
dt = DecisionTreeClassifier(random_state=28)

# Define the pipeline with scaling
pipeline_dt = Pipeline([
    ('scaler', StandardScaler()),
    ('decisionTreeClassifier', dt)
])

# Code Execution Confirmation
print("Decision Tree Successfully Initated and Pipeline Established")
```

<h3>9.2.2&nbsp;&nbsp;&nbsp;&nbsp;Decision Tree's KFold</h3>

<p>I will be using the KFold to conduct the cross-validation strategy within the GridSearchCV and RandomizedSearchCV to find the best Hyperparameter for the Decision Tree. KFold is also capable of reducing variance and bias which will directly improve the model's performance in terms of its fitting.</p>

```python
# Defining KFold Cross-Validator
kf = KFold(n_splits = 10, shuffle = True, random_state = 28)

# Code Execution Confirmation
print("KFold have been defined successfully")
```

<h3>9.2.3&nbsp;&nbsp;&nbsp;&nbsp;Decision Tree's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best Hyperparameter for the Decision Tree, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the Decision Tree. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good accuracy score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my Decision Tree, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid for GridSearchCV
grid_param_dt = {
    'decisionTreeClassifier__max_depth': [3, 5, 7, 9, None],
    'decisionTreeClassifier__min_samples_split': [20, 30, 40, 50],
    'decisionTreeClassifier__min_samples_leaf': [10, 15, 20, 25, 30],
    'decisionTreeClassifier__criterion': ['gini', 'entropy'],
    'decisionTreeClassifier__max_features': ['sqrt', 'log2', None]
}

# Create a GridSearchCV instance using KFold
grid_search_dt = GridSearchCV(
    estimator = pipeline_dt,
    param_grid = grid_param_dt,
    cv = kf,
    verbose = 1,
    n_jobs =- 1
)

# Fit GridSearchCV to the training data
grid_search_dt.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_dt.cv_results_)

# Sort the DataFrame to push the best parameter row to the top
grid_sorted_results_df = grid_results_df.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

This set of Parameter will be subsequently used to train the Decision Tree model and placed in comparison to the Parameter returned by GridSearchCV in the previous cell.</p>

<h3>10.2.3&nbsp;&nbsp;&nbsp;&nbsp;Logistics Regression's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best hyperparameter for the Logistics Regression, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the Logistics Regression. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good accuracy score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid to search for GridSearchCV
grid_param_lr = {
    'logisticRegression__C': np.logspace(-3, 3, 15),
    'logisticRegression__penalty': ['l1', 'l2', 'elasticnet'],
    'logisticRegression__solver': ['liblinear', 'saga'],
    'logisticRegression__class_weight': ['balanced', None],
    'logisticRegression__l1_ratio': [0.1, 0.5, 0.9]
}

# Create a GridSearchCV instance
grid_search_lr = GridSearchCV(
    estimator = pipeline_lr,
    param_grid = grid_param_lr,
    cv = kf,
    verbose = 1,
    n_jobs =- 1
)

# Fit GridSearchCV to the training data
grid_search_lr.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV
grid_results_df = pd.DataFrame(grid_search_lr.cv_results_)

grid_sorted_results_df = grid_results_df.sort_values(by = 'rank_test_score').reset_index(drop = True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

This Parameter will then be used to train the Logistics Regression model to then visualise the Learning Curve to see the performance of the Parameter. The result will also be compared to the Best Parameter that the RandomSearchCV return.</p>

RandomizedSearchCV is a lot more efficient in finding the Best Hyperparameter for the Logistics Regression as it has lesser amount of total fits as compared to GridSearchCV. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter distribution to search
param_dist_lr = {
    'logisticRegression__C': np.linspace(0.001, 20, 100),
    'logisticRegression__penalty': ['l1', 'l2', 'elasticnet'],
    'logisticRegression__solver': ['saga'],
    'logisticRegression__class_weight': ['balanced', None],
    'logisticRegression__l1_ratio': np.linspace(0.1, 1.0, 10)
}

# Create a RandomizedSearchCV instance
random_search_lr = RandomizedSearchCV(
    estimator = pipeline_lr,
    param_distributions = param_dist_lr,
    n_iter = 50,
    scoring = 'accuracy',
    cv = kf,
    random_state = 28,
    n_jobs =- 1,
    verbose = 2
)

# Fit RandomizedSearchCV to the training data
random_search_lr.fit(X_train, Y_train)

# Create a DataFrame to display the results
random_results_df = pd.DataFrame(random_search_lr.cv_results_)
random_sorted_results_df = random_results_df.sort_values(by = 'rank_test_score').reset_index(drop = True)
random_sorted_results_df.index += 1
highlighted_random_sorted_results_df = random_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_random_sorted_results_df)
```

<hr/>
<h3>10.3.1&nbsp;&nbsp;&nbsp;&nbsp;Gradient Boosting Classifier Initiation and Pipeline</h3>

<p>In order to start training the data for Gradient Boosting Classifier, I will firstly initiate it and also set up the pipeline which will scale the training data. I will also need to ensure that the random state is always 28 which is standardised throughout all the random state value. Scaling of data will also reduce biasness, ensure the feature is dimensionless and improve the model more accurate and effective as extreme values are removed.</p>

```python
# Create a Random Forest Classifier model
gbc = GradientBoostingClassifier(random_state = 28)

# Define the pipeline
pipeline_gbc = Pipeline([
    ('scaler', StandardScaler()),
    ('gradBoostClassifier', gbc)
])

# Code Execution Confirmation
print("Gradient Boosting Classifier Successfully Initated and Pipeline Established")
```

<h3>10.3.2&nbsp;&nbsp;&nbsp;&nbsp;Gradient Boosting Classifier KFold</h3>

<p>I will be using the KFold to conduct the cross-validation strategy within the GridSearchCV and RandomizedSearchCV to find the best hyperparameter for the Gradient Boosting Classifier. KFold is also capable of reducing variance and bias which will directly improve the model's performance in terms of its fitting.</p>

```python
# Defining KFold Cross-Validator
kf = KFold(n_splits = 5, shuffle = True, random_state = 28)

# Code Execution Confirmation
print("KFold have been defined successfully")
```

<h3>10.3.3&nbsp;&nbsp;&nbsp;&nbsp;Gradient Boosting Classifier's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best hyperparameter for the Gradient Boosting Classifier, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the Gradient Boosting Classifier. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good accuracy score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid for GridSearchCV
grid_param_gbc = {
    'gradBoostClassifier__n_estimators': [100, 150, 200],
    'gradBoostClassifier__learning_rate': [0.0005, 0.001, 0.005],
    'gradBoostClassifier__max_depth': [2, 3, 4],
    'gradBoostClassifier__min_samples_split': [40, 50, 60],
    'gradBoostClassifier__min_samples_leaf': [20, 25, 30],
    'gradBoostClassifier__subsample': [0.6, 0.7, 0.8],
    'gradBoostClassifier__max_features': ['sqrt', 'log2', None]
}

# Create a GridSearchCV instance using KFold for Gradient Boosting Classifier
grid_search_gbc = GridSearchCV(
    estimator = pipeline_gbc,
    param_grid = grid_param_gbc,
    cv = kf,
    verbose = 1,
    n_jobs =- 1
)

# Fit GridSearchCV to the training data for Gradient Boosting Classifier
grid_search_gbc.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV for Gradient Boosting Classifier
grid_results_df_gbc = pd.DataFrame(grid_search_gbc.cv_results_)

# Sort the DataFrame to push the best parameter row to the top
grid_sorted_results_df = grid_results_df_gbc.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df.index += 1
highlighted_grid_sorted_results_df = grid_sorted_results_df.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df)
```

This Parameter will then be used to train the Gradient Boosting Classifier model to then visualise the Learning Curve to see the performance of the Parameter. <h3>10.4.1&nbsp;&nbsp;&nbsp;&nbsp;AdaBoost Classifier's Initiation and Pipeline</h3>

<p>In order to start training the data for AdaBoost Classifier, I will firstly initiate it and also set up the pipeline which will scale the training data. I will also need to ensure that the random state is always 28 which is standardised throughout all the random state value. Scaling of data will also reduce biasness, ensure the feature is dimensionless and improve the model more accurate and effective as extreme values are removed.</p>

```python
# AdaBoost Classifier Initiation
ada = AdaBoostClassifier(random_state = 28)

# Establishing Pipeline for AdaBoost Classifier
pipeline_ada = Pipeline([
    ('scaler', StandardScaler()),
    ('adaBoostClassifier', ada)
])

# Code Execution Confirmation
print("AdaBoost Classifier Successfully Initated and Pipeline Established")
```

<h3>10.4.2&nbsp;&nbsp;&nbsp;&nbsp;AdaBoost Classifier's KFold</h3>

<p>I will be using the KFold to conduct the cross-validation strategy within the GridSearchCV and RandomizedSearchCV to find the best hyperparameter for the AdaBoost Classifier. KFold is also capable of reducing variance and bias which will directly improve the model's performance in terms of its fitting.</p>

```python
# Defining KFold Cross-Validator
kf = KFold(n_splits = 5, shuffle = True, random_state = 28)

# Code Execution Confirmation
print("KFold have been defined successfully")
```

<h3>10.4.3&nbsp;&nbsp;&nbsp;&nbsp;AdaBoost Classifier's Best Hyperparameter (Grid Search)</h3>

<p>In order to find the best hyperparameter for the AdaBoost Classifier, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the AdaBoost Classifier. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good accuracy score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid for GridSearchCV
grid_param_ada = {
    'adaBoostClassifier__n_estimators': [50, 75, 100],
    'adaBoostClassifier__learning_rate': [0.01, 0.05, 0.1]
}

# Create a GridSearchCV instance using KFold for AdaBoost Classifier
grid_search_ada = GridSearchCV(
    estimator=pipeline_ada,
    param_grid=grid_param_ada,
    cv=kf,
    verbose=1,
    n_jobs=-1
)

# Fit GridSearchCV to the training data for AdaBoost Classifier
grid_search_ada.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV for AdaBoost Classifier
grid_results_df_ada = pd.DataFrame(grid_search_ada.cv_results_)

# Sort the DataFrame to push the best parameter row to the top
grid_sorted_results_df_ada = grid_results_df_ada.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df_ada.index += 1
highlighted_grid_sorted_results_df_ada = grid_sorted_results_df_ada.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df_ada)
```

<p>With reference to the result returned by the GridSearchCV in the cell above, I can infer that the Best Parameter that GridSearchCV has returned is the Parameter that is displayed as the first index of the DataFrame as it hold a 'rank_test_score' of '1' which means it has performed much better compared to the other combination of Parameter. This Parameter will then be used to train the AdaBoost Classifier model to then visualise the Learning Curve to see the performance of the Parameter. The result will also be compared to the Best Parameter that the RandomSearchCV return.</p>

<h3>10.5.1&nbsp;&nbsp;&nbsp;&nbsp;K Neighbors Classifier Initiation and Pipeline</h3>

<p>In order to start training the data for K Neighbors Classifier, I will firstly initiate it and also set up the pipeline which will scale the training data.
```python
# Initation of K Neighbors Classifier
knn = KNeighborsClassifier()

# Establishing Pipeline for K Neighbors Classifier
pipeline_knn = Pipeline([
    ('scaler', StandardScaler()),
    ('kNeighborClassifier', knn)
])

# Code Execution Confirmation
print("K Neighbors Classifier Successfully Initated and Pipeline Established")
```

KFold is also capable of reducing variance and bias which will directly improve the model's performance in terms of its fitting.</p>

```python
# Establishing KFold for Support Vector Classifier
kf = KFold(n_splits = 5, shuffle = True, random_state = 28)

# Code Execution Confirmation
print("KFold have been defined successfully")
```

<h3>10.5.3&nbsp;&nbsp;&nbsp;&nbsp;K Neighbors Classifier's Best Parameter (Grid Search)</h3>

<p>In order to find the best hyperparameter for the K Neighbors Classifier, I will need to run GridSearchCV which will search through all my parameter combination and fit it to the training data to verify which parameter produced the best result for the K Neighbors Classifier. GridSearchCV is the most detailed way to find the best Hyperparameter but it will be at the expense of time as GridSearchCV takes up tremendous amount of time due to it iterating through every possible combination of Hyperparameter. The best result is defined by the following criteria,</p>
<br/>
<p><strong><u>Accuracy</u></strong></p>
<ul>
    <li>A good Hyperparameter should produce a good accuracy score that can be verified on the Learning Curve.</li>
</ul>
<p><strong><u>Ideal Fitting</u></strong></p>
<ul>
    <li>A good Hyperparameter should neither be overfitting nor underfitting. Instead the Cross-Validation Score should be relatively in close proximity with the Training Score on the Learning Curve.</li>
</ul>
<br/>
<p>However, despite wanting the best Hyperparameter for my model, there should be a balance of both Accuracy and Ideal Fitting as to address fitting to mitigate Bias and Variance will affect the accuracy of the model and to address the Accuracy of the model will affect the amount of Bias and Variance present. Therefore, a perfect balance needs to be found between these 2 criteria.</p>

```python
# Define the hyperparameter grid for GridSearchCV
grid_param_knn = {
    'kNeighborClassifier__n_neighbors': [51, 75, 101],
    'kNeighborClassifier__weights': ['uniform'],
    'kNeighborClassifier__algorithm': ['auto', 'kd_tree'],
    'kNeighborClassifier__leaf_size': [10, 20, 30],
    'kNeighborClassifier__p': [1]
}

# Create a GridSearchCV instance using KFold for K-Nearest Neighbors
grid_search_knn = GridSearchCV(
    estimator=pipeline_knn,
    param_grid=grid_param_knn,
    cv=kf,
    verbose=1,
    n_jobs=-1
)

# Fit GridSearchCV to the training data for K-Nearest Neighbors
grid_search_knn.fit(X_train, Y_train)

# Create a DataFrame to display the results of GridSearchCV for K-Nearest Neighbors
grid_results_df_knn = pd.DataFrame(grid_search_knn.cv_results_)

# Sort the DataFrame to push the best parameter row to the top
grid_sorted_results_df_knn = grid_results_df_knn.sort_values(by='rank_test_score').reset_index(drop=True)
grid_sorted_results_df_knn.index += 1
highlighted_grid_sorted_results_df_knn = grid_sorted_results_df_knn.head(3).style.apply(lambda x: ['background-color: darkgreen' if x.name == 1 else '' for _ in x], axis=1).set_properties(**{'border': '1px solid black'})
display(highlighted_grid_sorted_results_df_knn)
```

<p>With reference to the result returned by the GridSearchCV in the cell above, I can infer that the Best Parameter that GridSearchCV has returned is the Parameter that is displayed as the first index of the DataFrame as it hold a 'rank_test_score' of '1' which means it has performed much better compared to the other combination of Parameter. <p>To define the Voting Classifier, I will have to input all the previous models that are intended for the use in Voting Classifier. Afterwards, I will need to select the voting method that will be utilised. There are two different voting methods that can be used. The details regarding the voting methods are stated below.</p>
<br/>
<p><strong><u>Hard Voting</u></strong></p>
<ul>
    <li>Each model within the Voting Classifier votes for a target, and the final prediction is made based on majority votes.</li>
    <li>Advantage: Easy implementation.</li>
    <li>Disadvantage: Confidence level of each prediction is not taken into account.</li>
</ul>
<p><strong><u>Soft Voting</u></strong></p>
<ul>
    <li>Collects the predicted probability from each model and calculate the average to make the prediction</li>
    <li>Advantage: Higher accuracy as confidence level is taken into account.</li>
    <li>Disadvantage: Requires models that can predict probability.</li>
</ul>
<br/>
<p>With the evaluation and details of the two voting methods listed above, I have decided to proceed with Soft Voting in light of its accuracy compared to Hard Voting. I will proceed with defining and choosing the method of voting in the cell below.</p>

<h2>11.5&nbsp;&nbsp;&nbsp;&nbsp;Voting Classifier's Threshold Improvement</h2>

<p>Due to class inbalance as there are more Normal Machine Status data compared to Abnormal Machine Status data, the adjustment of threshold is required to mitigate this inbalance. The looping and visualisation of a range of value is needed to achieve the desired Threshold value. The default Threshold is 0.5, thus the for loop will start from 0.1 and up to 1.0 in the steps of 0.00005. This operation will be conducted in the cell below.</p>

```python
# Fit the Voting Classifier to the training data
voting_clf.fit(X_train, Y_train)

# Predict probabilities for the positive class
y_probs = voting_clf.predict_proba(X_test)[:, 1]

# Define the range of thresholds to evaluate
thresholds = np.arange(0.1, 1.0, 0.00005)
best_threshold = 0
best_f1_score = 0

# List to store F1 scores for each threshold
threshold_metrics = []

# Loop through the threshold values
for threshold in thresholds:
    # Make predictions based on current threshold
    y_pred_adjusted = (y_probs >= threshold).astype(int)

    # Calculate F1 Score for the current threshold
    f1 = f1_score(Y_test, y_pred_adjusted, average='weighted')

    # Store the threshold and corresponding F1 Score
    threshold_metrics.append({"Threshold": threshold, "F1 Score": f1})

    # Update the best threshold if this one is better
    if f1 > best_f1_score:
        best_f1_score = f1
        best_threshold = threshold

# Convert the metrics into a DataFrame for easier visualization
threshold_df = pd.DataFrame(threshold_metrics)

# Display the best threshold and F1 score in a separate DataFrame
best_threshold_df = pd.DataFrame({
    "Best Threshold": [best_threshold],
    "Best F1 Score": [best_f1_score]
}, index = ["1"])

# Plot the F1 Score vs Threshold using seaborn
plt.figure(figsize=(10, 6))
sns.lineplot(data=threshold_df, x = "Threshold", y = "F1 Score", color = 'b')
plt.xlabel('Threshold')
plt.ylabel('F1 Score')
plt.title('F1 Score vs Threshold')
plt.grid(True)
plt.show()

# Display the DataFrame with the best threshold and F1 Score
best_threshold_df
```

<p>With reference to the Graph and DataFrame above, I am able to visually verify that the peak of the curve which is the ideal Threshold is around 0.3 as that is where the F1 Score is also at the highest. Subsequently, the DataFrame also provides the numerical value for the ideal Threshold. This Threshold will then be used for the Voting Classifier.</p>

<h2>12.1&nbsp;&nbsp;&nbsp;&nbsp;Deployment of Model</h2>

<p>The Deployment of the model in a factory environment can be done through multiple ways. However each method of deployment and integration has its respective advantages and disadvantages. Some of these factors to consider before making a decision on how to deploy the model are listed below.</p>
<br/>
<p><strong><u>Basis of Need and Infrastructure Requirements</u></strong></p>
<ul>
    <li>Should a prediction require a Real-Time need, there must be low-latency during the transmition and receiving of the Real-Time data which require higher standards of Infrastructure Requirements.</li>
    <li>Reliable predictions must be provided if the factory intent to apply immediate corrections / rectification for the machine.</li>
</ul>
<p><strong><u>Scale of Deployment and Financial Implications</u></strong></p>
<ul>
    <li>Deploying any AIML models in large scale to ensure coverage for all machines will require large amount of Financial Resources.</li>
    <li>Deploying any AIML models at a small scale will create less strain on Financial Resources but will affect the overall coverage of machines.</li>
    <li>Some reasons for large Financial Implications may include high usage of servers and installation of IoT systems (e.g. Scanners) within machines to provide data for predictions.</li>
</ul>
<p><strong><u>User Accessibility</u></strong></p>
<ul>
    <li>Staff within the factory must be trained on the usage of the AIML model.</li>
    <li>Dedicated Staff and Worker must also be deployed to observe the predictions and quickly address the machine issues if it is predicted to be abnormal.</li>
    <li>For future training data, Staff and Worker must also record down the down a machine's status and data features frequently to allow future training and to keep the AIML model up to date and shine light in new relationship of features to the target.</li>
</ul>
<p><strong><u>Risk Assessment and Management</u></strong></p>
<ul>
    <li>All AIML models are prone to Incorrect Predictions. These incorrect predictions can cause wastage of resources (e.g. Manpower) and potentially financial implications.</li>
    <li>The factory will have to decide if integrating AIML models into its workflow is ideal in view of potential risk.</li>
</ul>
<br/>
<p>The factors listed above are just some factors to take into account for a factory when they wish to deploy a AIML model into the factory setting. Listed in point form below are some ways the AIML models can be deployed in a factory environment.</p>
<br/>
<p><strong><u>On-site Deployment</u></strong></p>
<ul>
    <li>Deploy the AIML model within the Factory with local servers within the premise.</li>
    <li>Advantage: Real-time predictions and very little latency / delay.</li>
    <li>Disadvantage: Extremely high financial cost and requires a IT Team to maintain the servers.</li>
    <li>Not advisable for Factories with small-scale operation and low revenue.</li>
</ul>
<p><strong><u>Cloud Deployment</u></strong></p>
<ul>
    <li>Host the AIML model on the Cloud using reliable Cloud Service Provider (e.g. Amazon Web Service).</li>
    <li>Advantage: Low cost of implementation and can be used at different scales.</li>
    <li>Disadvantage: Performance is reliant on network connectivity and may not be very efficient.</li>
    <li>Not advisable for Factories located in regions with low network connectivity.</li>
    <li>Advisable for Factories that are operating at a small-scale and with low revenue.</li>
</ul>
<p><strong><u>Edge Deployment</u></strong></p>
<ul>
    <li>Deploy the AIML model on Edge Devices located near the machines.</li>
    <li>Advantage: Low latency with limited delays and capable of operating without network.</li>
    <li>Disadvantage: Limited computational capabilities and require specialised hardwares.</li>
    <li>Not advisable for Factories that operate in extremely large capacity.</li>
    <li>Advisable for Factories located in regions with low network connectivity.</li>
</ul>
<br/>
<p>The potential methods of deploying a AIML model has been stated above. Each method of deployment has its fair share of advantages and disadvantages. Therefore, it is up to each individual factory to decide on which approach of deployment they wish to adopt as each factories have insights to their respective operational capacity and resources.</p>
